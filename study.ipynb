```python
The perceptron is a fundamental component of many neural nets.
The poerceptron was modeled after a neuron and while there is a limit to what can be done with a single perceptron 
(well highlight this later) a single perceptron can be used for simple binary classification.
This classification will corresponded to the model firing or not firing for a given sample.
```

```python
Mathematically, the output of a perceptron is modeled as:

f(x₁·w₁ + x₂·w₂ + ... + xₙ·wₙ + b) = y
x₁, x₂, ..., xₙ are the inputs
w₁, w₂, ..., wₙ are the weights
b is the bias (similar to a neuron's firing threshold)
f is the activation function (e.g., step function: fire or don’t fire)
y is the final output
```
